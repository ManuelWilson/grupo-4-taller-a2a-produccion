### 4.4 API de A2A

Con `a2a-sdk 1.2.3` la API real es:

| Antes (no existe en 1.2.3) | Ahora |
|---|---|
| `from a2a.client.transports import TransportProtocol` | `ClientConfig(supported_protocol_bindings=["JSONRPC"])` |
| `MessageSendParams`, `TextPart`, `Role.user` | `SendMessageRequest(message=new_text_message(...))` |
| `create_client(card, config=...)` | `await create_client(card, client_config=...)` |
| `agent_card.model_dump(...)` | `MessageToDict(agent_card)` (la Agent Card es protobuf) |

Además, el servidor responde en formato A2A 1.0 y el proto de `a2a-sdk` no lo parsea: se fija
`interfaz.protocol_version = "0.3.0"` antes de `create_client`. La Agent Card no expone `.url`;
el endpoint se lee de `supported_interfaces`.
